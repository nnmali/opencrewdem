```python

from flask import Flask, jsonify, request

app = Flask(__name__)

team = [
    {"id": 1, "name": "Alice"},
    {"id": 2, "name": "Bob"}
]

team
```

```python
print("hello world")
```

```python

@app.route("/")
def home():
    return jsonify({"message": "Welcome to OpenCrew"})
```

```python


@app.route("/health")
def health():
    return jsonify({"status": "healthy"})

```

```python

@app.route("/team", methods=["GET"])
def get_team():
    return jsonify(team)


```

```python
@app.route("/team", methods=["POST"])
def add_team_member():
    data = request.get_json()

    member = {
        "id": len(team) + 1,
        "name": data["name"]
    }

    team.append(member)

    return jsonify(member), 201

```

```python

if __name__ == "__main__":
    app.run(debug=True)
```
